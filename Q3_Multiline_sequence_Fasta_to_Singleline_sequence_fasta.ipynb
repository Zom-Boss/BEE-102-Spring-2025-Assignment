Multi-line sequence fasta to single-line fasta file

```python
def Converting(inpFile, outFile):
    # Opening the file for reading and writing
    with open(inpFile, "r") as infile, open(outFile, "w") as outfile:
        Header = None     # For headers
        Sequence = []     # For storing sequences

        # Reading begins :
        for lines in infile:
            lines = lines.strip()  # Removing the spaces

            if lines.startswith(">"):  # If header is found
                if Header:
                    outfile.write(f"{Header}\n{''.join(Sequence)}\n")

                Header = lines
                Sequence = []  # Reseting for new header
            else:
                Sequence.append(lines) # Just concatenate

        if Header:
            outfile.write(f"{Header}\n{''.join(Sequence)}\n")


# I am defining the input and output file
inpFile = "multiline_input.fasta.txt"
outFile = "ans.txt"


Converting(inpFile,outFile)
from google.colab import files
files.download("ans.txt")
# You can now see the output in ans.txt file
# The file will be downloaded automatically
```
