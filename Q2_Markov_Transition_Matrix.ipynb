```python
import pandas as pd

def funC(seq):
    bases = ['A', 'C', 'G', 'T']

    # I have made a Dictionary for counting every possible transitions
    counts = {a1: {a2: 0 for a2 in bases} for a1 in bases}

    # Conting the transitions by looping on the sequence
    for i in range(len(seq) - 1):
        cur = seq[i]
        nxt = seq[i + 1]
        if cur in bases and nxt in bases:
            counts[cur][nxt] += 1

    df = pd.DataFrame(counts).T
    matrix = df.div(df.sum(axis=1), axis=0)
    return matrix

# I have taken a randomly generated sequence of length 60
#  We can also take the input from the user
seq = "AGTCAGTCCAGTACGATGCAATGCGATCGATCGATCGTTAGCTAGCTAGCTAG"

matrix = funC(seq)
print(matrix)
```
